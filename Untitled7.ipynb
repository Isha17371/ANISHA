```python
import math
# Settlement Calculation

C = 0.24  #Compression index
H = 3   #Thickness of compressible layer (m)
e = 0.8  #Initial void ratio
p =  30 #Initial effective stress (Kn/m2)
p1 = 124.60 #Stress increase due to load (Kn/m2)

S = (C * H / (1 + e)) * math.log10((p + p1) / p)

print("Settlement = ", round(S, 2), "mm ", "= ", end = "")
print(round(S*1000,2), "m")
```
