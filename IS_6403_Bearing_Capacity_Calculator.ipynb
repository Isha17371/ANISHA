```python
import math

# Settlement Calculation
# -------- INPUTS --------
C = 0.24  #Compression index
H = 3   #Thickness of compressible layer (m)
e = 0.8  #Initial void ratio
p =  30 #Initial effective stress (Kn/m2)
p1 = 124.60 #Stress increase due to load (Kn/m2)

S = (C * H / (1 + e)) * math.log10((p + p1) / p)

print("Settlement = ", round(S, 2), "mm ", "= ", end = "")
print(round(S*1000,2), "m")

# TERZAGHI BEARING CAPACITY
# -------- INPUTS --------

B = 1.5          # Width of footing (m)
Df = 1.0         # Depth of foundation (m)
FOS = 3
gamma = 18.0     # Unit weight of soil (kN/m3)
c = 15.0         # Cohesion (kPa)
phi = 20.0       # Angle of internal friction (degrees)

# Terzaghi bearing capacity factors
Nc = 17.7
Nq = 7.4
Ngamma = 5.0
# Shape Factor
sc = 1
sq = 1
sgamma = 1
# Inclination Factor
ic = 1
iq = 1
igamma = 1
# Depth Factor
dc = 1
dq = 1
dgamma = 1

# -------- SURCHARGE --------
q = gamma * Df

# Gross Ultimate Bearing Capacity
qu = (c * Nc * sc * dc * ic + q * Nq * sq * dq * iq + 0.5 * gamma * B * Ngamma * sgamma * dgamma * igamma)

# Net Ultimate Bearing Capacity
qnu = (qu - q)

# Net Safe Bearing Capacity
qns = (qnu / FOS)

# Gross Safe Bearing Capacity
qs = (qns + q)

print("Gross Ultimate Bearing Capacity =", round(qu,2), "kPa")
print("Net Ultimate Bearing Capacity   =", round(qnu,2), "kPa")
print("Net Safe Bearing Capacity       =", round(qns,2), "kPa")
print("Gross Safe Bearing Capacity     =", round(qs,2), "kPa")



```
